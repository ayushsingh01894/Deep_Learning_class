```python
import numpy as np
import pandas as pd
import keras
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import keras.models as km
from keras.models import Sequential
from keras import activations, initializers, regularizers, constraints
from keras.layers import Dense, Activation
```

```python
dataset = pd.read_csv("concrete_strength.csv")
```

```python
dataset
```

```python
dataset.describe()
```

```python
X = dataset.drop(columns=["compressive_strength"])
y = dataset["compressive_strength"]
```

```python
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.3, random_state=123)
```

```python
sc = StandardScaler()
train_X = sc.fit_transform(train_X)
test_X = sc.transform(test_X)
```

```python
train_X.shape
```

```python
model = Sequential()
model.add(Dense(units=8, kernel_initializer='uniform', activation='relu', input_dim=8))
model.add(Dense(units=6, kernel_initializer='uniform', activation='relu'))
model.add(Dense(units=4, kernel_initializer='uniform', activation='relu'))
model.add(Dense(units=1, kernel_initializer='uniform', activation='linear'))
```

```python
model.summary()
```

```python
model.compile(loss='mean_squared_error', optimizer='adam')
```

```python
print(train_X.shape)
print(train_y.shape)
```

```python
history = model.fit(train_X, train_y, validation_data=(test_X, test_y), epochs=50, batch_size=32)
```

```python
predictions = model.predict(test_X)
predictions
```

```python
from sklearn.metrics import mean_squared_error, r2_score
print("MSE:", mean_squared_error(test_y, predictions))
print("R2 Score:", r2_score(test_y, predictions))
```

```python
import pickle

# Model save karein
model.save("concrete_model.h5")

# Scaler (sc) ko pickle me save karein
with open("scaler.pkl", "wb") as f:
    pickle.dump(sc, f)

print("Model aur Scaler dono successfully save ho gaye!")
```
